```python
import pandas as pd 
import numpy as np
import plotly.express as px

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler,StandardScaler
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error,root_mean_squared_error
from sklearn.pipeline import make_pipeline
```

```python
from sklearn.datasets import make_regression, fetch_california_housing
# 1. Create a tricky dataset with 80 features, but only 5 are actually useful!
X_data, y_data = make_regression(n_samples=100, n_features=80, n_informative=5, noise=20, random_state=42)

# Convert to a Pandas DataFrame so it's easy to read
feature_names = [f"Feature_{i}" for i in range(1, 81)]
X = pd.DataFrame(X_data, columns=feature_names)
y = pd.Series(y_data, name="House_Price")

print("--- OUR TRICKY DATASET ---")
print(f"We have {X.shape[0]} houses and {X.shape[1]} features (clues).")
display(X.head())
```

as there are too mmany feaures buttt only some of them are just the garbage . tthe solution of it is the regularization as it will regulate the importance of the feature 
1. ridge regression
2. Lasso regression

```python
# Spliting the  data
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)

# Scaling the data
scaler=MinMaxScaler()
X_train_Scaled=scaler.fit_transform(X_train)
X_test_Scaled=scaler.transform(X_test)

# training the data
model=LinearRegression()
model.fit(X_train_Scaled,y_train)

# printing the training andd testing data
print("Training score : ",model.score(X_train_Scaled,y_train))
print("testing score : ",model.score(X_test_Scaled,y_test))
```

As we can see that with the useless features are overtraining the model and having very less testing score as compared to training score . So we first gonna try the ridge on it 
### Ridge : 
A regularized version of linear regression that adds a penalty proportional to the sum of the squared values of all coefficients, forcing them to stay small but never reducing any to exactly zero.


```python
model_r=Ridge(alpha=0.1)
model_r.fit(X_train_Scaled,y_train)
#  printing the ridge scores
print("Ridge training score : ",model_r.score(X_train_Scaled,y_train))
print("Ridge testing score  : ",model_r.score(X_test_Scaled,y_test))
```

Now as we tried the ridge on the data and got better results on same over fitted data . But there is one other better option called lasso . As ridge just penalize the features , on the other hand Lasso put those noices to zero .

Lasso : A regularized version of linear regression that adds a penalty proportional to the sum of the absolute values of all coefficients — this can reduce some coefficients all the way to exactly zero, effectively removing those features from the model.

```python
model_l=Lasso(alpha=0.2)
model_l.fit(X_train_Scaled,y_train)

print("Lasso training score : ",model_l.score(X_train_Scaled,y_train))
print("Lasso Testing Score : ",model_l.score(X_test_Scaled,y_test))
```
